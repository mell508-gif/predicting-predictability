```python
from rdkit import Chem

def canonicalize_smiles(smiles):
    if pd.isna(smiles):
        return None
    
    mol = Chem.MolFromSmiles(smiles)
    return Chem.MolToSmiles(mol) if mol is not None else None

tahoe_compounds["canonical_smiles"] = tahoe_compounds["smiles"].apply(canonicalize_smiles)

print("Valid canonical Tahoe SMILES:", tahoe_compounds["canonical_smiles"].notna().sum())
print("Invalid/unreadable Tahoe SMILES:", tahoe_compounds["canonical_smiles"].isna().sum())
```

```python
# Canonicalize existing registry SMILES
existing_registry["canonical_smiles"] = existing_registry["smiles"].apply(
    canonicalize_smiles
)

# Canonicalize DRUG-seq SMILES
drugseq_usable_compounds["canonical_smiles"] = (
    drugseq_usable_compounds["smiles"].apply(canonicalize_smiles)
)

# Make sets of structures already represented
existing_smiles = set(
    existing_registry["canonical_smiles"].dropna()
)

drugseq_smiles = set(
    drugseq_usable_compounds["canonical_smiles"].dropna()
)

# Label Tahoe coverage
tahoe_registry_status = tahoe_compounds.copy()

tahoe_registry_status["in_existing_registry"] = (
    tahoe_registry_status["canonical_smiles"].isin(existing_smiles)
)

tahoe_registry_status["in_drugseq"] = (
    tahoe_registry_status["canonical_smiles"].isin(drugseq_smiles)
)

tahoe_registry_status["new_to_both"] = (
    tahoe_registry_status["canonical_smiles"].notna()
    & ~tahoe_registry_status["in_existing_registry"]
    & ~tahoe_registry_status["in_drugseq"]
)

print("Tahoe compounds with valid structures:",
      tahoe_registry_status["canonical_smiles"].notna().sum())

print("Already in existing 6,801 registry:",
      tahoe_registry_status["in_existing_registry"].sum())

print("Already in DRUG-seq:",
      tahoe_registry_status["in_drugseq"].sum())

print("New to both:",
      tahoe_registry_status["new_to_both"].sum())
```

```python
# Build expanded compound structure registry
```

```python
# Keep only existing-registry compounds with usable canonical SMILES
existing_for_registry = existing_registry[
    existing_registry["canonical_smiles"].notna()
].copy()

existing_for_registry = existing_for_registry.rename(
    columns={
        "name": "compound_name",
        "smiles": "original_smiles",
    }
)

existing_for_registry["source_existing_registry"] = True
existing_for_registry["source_drugseq"] = False
existing_for_registry["source_tahoe"] = False
```

```python
# Add canonical SMILES to the DRUG-seq overlap/status table
drugseq_registry_status = drugseq_registry_status.copy()

drugseq_registry_status["canonical_smiles"] = (
    drugseq_registry_status["smiles"].apply(canonicalize_smiles)
)

# Keep only compounds with readable structures
drugseq_for_registry = drugseq_registry_status[
    drugseq_registry_status["canonical_smiles"].notna()
].copy()

# Rename columns to match the shared expanded-registry format
drugseq_for_registry = drugseq_for_registry.rename(
    columns={
        "name": "compound_name",
        "smiles": "original_smiles",
    }
)

# Source flags
drugseq_for_registry["source_existing_registry"] = (
    drugseq_for_registry["already_in_existing_registry"]
)

drugseq_for_registry["source_drugseq"] = True
drugseq_for_registry["source_tahoe"] = False

# Preview only the columns we will eventually combine
drugseq_for_registry[
    [
        "compound_name",
        "original_smiles",
        "canonical_smiles",
        "source_existing_registry",
        "source_drugseq",
        "source_tahoe",
    ]
].head()
```

```python
# Keep Tahoe compounds with usable canonical SMILES
tahoe_for_registry = tahoe_registry_status[
    tahoe_registry_status["canonical_smiles"].notna()
].copy()

tahoe_for_registry = tahoe_for_registry.rename(
    columns={
        "drug": "compound_name",
        "smiles": "original_smiles",
    }
)

tahoe_for_registry["source_existing_registry"] = (
    tahoe_for_registry["in_existing_registry"]
)

tahoe_for_registry["source_drugseq"] = tahoe_for_registry["in_drugseq"]
tahoe_for_registry["source_tahoe"] = True

tahoe_for_registry[
    [
        "compound_name",
        "original_smiles",
        "canonical_smiles",
        "source_existing_registry",
        "source_drugseq",
        "source_tahoe",
    ]
].head()
```

```python
registry_columns = [
    "compound_name",
    "original_smiles",
    "canonical_smiles",
    "source_existing_registry",
    "source_drugseq",
    "source_tahoe",
]

expanded_registry_raw = pd.concat(
    [
        existing_for_registry[registry_columns],
        drugseq_for_registry[registry_columns],
        tahoe_for_registry[registry_columns],
    ],
    ignore_index=True,
)

print("Rows before removing duplicate structures:", len(expanded_registry_raw))
expanded_registry_raw.head()
```

```python
expanded_registry = (
    expanded_registry_raw
    .groupby("canonical_smiles", as_index=False)
    .agg(
        compound_name=("compound_name", "first"),
        original_smiles=("original_smiles", "first"),
        source_existing_registry=("source_existing_registry", "max"),
        source_drugseq=("source_drugseq", "max"),
        source_tahoe=("source_tahoe", "max"),
    )
)

print("Unique structures in expanded registry:", len(expanded_registry))
print("Present in original registry:", expanded_registry["source_existing_registry"].sum())
print("Present in DRUG-seq:", expanded_registry["source_drugseq"].sum())
print("Present in Tahoe:", expanded_registry["source_tahoe"].sum())

expanded_registry.head()
```

```python
expanded_registry.to_csv(
    Path.cwd() / "expanded_compound_structure_registry.csv",
    index=False
)

print("Saved expanded registry.")
```

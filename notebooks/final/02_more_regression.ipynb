```python
target_col = "source_normalized_inverted_score"

SCORE_PATH = (
    DATA_DIR
    / "step3_multimodal"
    / "step3_compound_predictability_latent256_final_scores.csv"
)

scores = pd.read_csv(SCORE_PATH)

print("Loaded:", SCORE_PATH)
print("Shape:", scores.shape)
print("Target exists:", target_col in scores.columns)

assert target_col in scores.columns, (
    f"{target_col} is missing. Available score columns: "
    f"{[col for col in scores.columns if 'score' in col.lower()]}"
)

required_columns = [
    "compound_embedding_row_index",
    "compound_name",
    "n_profiles",
    target_col,
]

compound_table = (
    scores
    .dropna(subset=required_columns)
    .copy()
)

compound_table["compound_embedding_row_index"] = (
    compound_table["compound_embedding_row_index"]
    .astype(int)
)

compound_table["n_profiles"] = (
    compound_table["n_profiles"]
    .astype(int)
)

compound_table = compound_table.reset_index(drop=True)

print("Original score rows:", len(scores))
print("Usable compound rows:", len(compound_table))
print(
    "Unique embedding indices:",
    compound_table["compound_embedding_row_index"].nunique()
)
print(
    "Duplicate embedding indices:",
    compound_table["compound_embedding_row_index"]
    .duplicated()
    .sum()
)

display(
    compound_table[
        [
            "compound_embedding_row_index",
            "compound_name",
            "n_profiles",
            target_col,
        ]
    ].head()
)

display(
    compound_table[
        ["n_profiles", target_col]
    ].describe()
)
```

```python
import numpy as np

EMBEDDING_PATH = (
    DATA_DIR
    / "step1_embeddings"
    / "step1_drug_embeddings_128.npy"
)

print("Embedding file:", EMBEDDING_PATH)
print("Exists:", EMBEDDING_PATH.exists())

# Avoid loading all 3.7 million
all_embeddings = np.load(
    EMBEDDING_PATH,
    mmap_mode="r"
)

print("Full embedding array shape:", all_embeddings.shape)
print("Embedding dtype:", all_embeddings.dtype)

if all_embeddings.ndim != 2:
    raise ValueError(
        f"Expected a 2D array, got shape {all_embeddings.shape}"
    )

if all_embeddings.shape[1] != 128:
    raise ValueError(
        f"Expected 128 latent dimensions, got {all_embeddings.shape[1]}"
    )

embedding_indices = (
    compound_table["compound_embedding_row_index"]
    .to_numpy(dtype=int)
)

print("Minimum requested index:", embedding_indices.min())
print("Maximum requested index:", embedding_indices.max())
print("Number of available embedding rows:", all_embeddings.shape[0])

if embedding_indices.min() < 0:
    raise IndexError("Negative embedding index detected.")

if embedding_indices.max() >= all_embeddings.shape[0]:
    raise IndexError(
        "At least one embedding index is outside the array."
    )

# Extract only the 6,605 profiled compounds
X_all = np.asarray(
    all_embeddings[embedding_indices],
    dtype=np.float32
)

y_all = compound_table[target_col].to_numpy(
    dtype=np.float32
)

print("\nX_all shape:", X_all.shape)
print("y_all shape:", y_all.shape)
print("All X values finite:", np.isfinite(X_all).all())
print("All y values finite:", np.isfinite(y_all).all())

assert X_all.shape == (len(compound_table), 128)
assert y_all.shape == (len(compound_table),)
assert np.isfinite(X_all).all()
assert np.isfinite(y_all).all()
```

```python
thresholds = [2, 5, 10, 20, 50, 100]

threshold_rows = []

for threshold in thresholds:
    subset = compound_table[
        compound_table["n_profiles"] >= threshold
    ]

    threshold_rows.append({
        "minimum_profiles": threshold,
        "n_compounds": len(subset),
        "mean_n_profiles": subset["n_profiles"].mean(),
        "median_n_profiles": subset["n_profiles"].median(),
        "mean_target": subset[target_col].mean(),
        "median_target": subset[target_col].median(),
        "target_std": subset[target_col].std(),
        "target_min": subset[target_col].min(),
        "target_max": subset[target_col].max(),
    })

threshold_summary = pd.DataFrame(threshold_rows)

display(threshold_summary)

print(
    "Compounds with at least 100 profiles:",
    int((compound_table["n_profiles"] >= 100).sum())
)
```

```python
from sklearn.ensemble import ExtraTreesRegressor
from sklearn.metrics import mean_absolute_error, r2_score
from sklearn.model_selection import RepeatedKFold
from scipy.stats import spearmanr

RANDOM_STATE = 42

# Thresholds
evaluation_thresholds = [
    5,
    10,
    20,
    25,
    50,
    75,
    100,
]

cv = RepeatedKFold(
    n_splits=5,
    n_repeats=3,
    random_state=RANDOM_STATE,
)

cv_rows = []

for threshold in evaluation_thresholds:
    mask = (
        compound_table["n_profiles"] >= threshold
    ).to_numpy()

    X = X_all[mask]
    y = y_all[mask]

    print(
        f"\nThreshold >= {threshold}: "
        f"{len(y)} compounds"
    )

    for fold_number, (train_idx, test_idx) in enumerate(
        cv.split(X),
        start=1,
    ):
        model = ExtraTreesRegressor(
            n_estimators=300,
            random_state=RANDOM_STATE + fold_number,
            n_jobs=-1,
        )

        model.fit(
            X[train_idx],
            y[train_idx],
        )

        predictions = model.predict(
            X[test_idx]
        )

        r2 = r2_score(
            y[test_idx],
            predictions,
        )

        mae = mean_absolute_error(
            y[test_idx],
            predictions,
        )

        spearman = spearmanr(
            y[test_idx],
            predictions,
        ).correlation

        if np.isnan(spearman):
            spearman = 0.0

        cv_rows.append({
            "minimum_profiles": threshold,
            "n_compounds": len(y),
            "fold": fold_number,
            "r2": r2,
            "mae": mae,
            "spearman": spearman,
        })

extra_trees_cv_results = pd.DataFrame(cv_rows)

print("\nFinished threshold sweep.")
display(extra_trees_cv_results.head())
```

```python
extra_trees_threshold_summary = (
    extra_trees_cv_results
    .groupby(
        [
            "minimum_profiles",
            "n_compounds",
        ],
        as_index=False,
    )
    .agg(
        mean_r2=("r2", "mean"),
        std_r2=("r2", "std"),
        mean_spearman=("spearman", "mean"),
        std_spearman=("spearman", "std"),
        mean_mae=("mae", "mean"),
        std_mae=("mae", "std"),
    )
    .sort_values(
        "minimum_profiles"
    )
    .reset_index(drop=True)
)

display(extra_trees_threshold_summary)
```

```python
from sklearn.ensemble import ExtraTreesRegressor
from sklearn.metrics import (
    mean_absolute_error,
    r2_score,
)
from sklearn.model_selection import RepeatedKFold
from scipy.stats import spearmanr

RANDOM_STATE = 42

evaluation_thresholds = [
    5,
    10,
    20,
    25,
    50,
    75,
    100,
]

feature_sets = {
    "128D_embedding_only": X_all,
    "130D_embedding_plus_step1_QC": X_all_130,
}

comparison_rows = []

for threshold in evaluation_thresholds:
    mask = (
        compound_table["n_profiles"] >= threshold
    ).to_numpy()

    y = y_all[mask]

    print("\n" + "=" * 90)
    print(
        f"Threshold >= {threshold}: "
        f"{len(y)} compounds"
    )

    # Identical splits
    cv = RepeatedKFold(
        n_splits=5,
        n_repeats=3,
        random_state=RANDOM_STATE,
    )

    splits = list(
        cv.split(np.arange(len(y)))
    )

    for feature_set_name, full_X in feature_sets.items():
        X = full_X[mask]

        print(f"  Feature set: {feature_set_name}")

        for fold_number, (train_idx, test_idx) in enumerate(
            splits,
            start=1,
        ):
            # Match the original successful Step 4 model settings
            model = ExtraTreesRegressor(
                n_estimators=400,
                max_depth=None,
                min_samples_leaf=3,
                random_state=RANDOM_STATE,
                n_jobs=-1,
            )

            model.fit(
                X[train_idx],
                y[train_idx],
            )

            predictions = model.predict(
                X[test_idx]
            )

            r2 = r2_score(
                y[test_idx],
                predictions,
            )

            mae = mean_absolute_error(
                y[test_idx],
                predictions,
            )

            spearman = spearmanr(
                y[test_idx],
                predictions,
            ).correlation

            if np.isnan(spearman):
                spearman = 0.0

            comparison_rows.append({
                "feature_set": feature_set_name,
                "minimum_profiles": threshold,
                "n_compounds": len(y),
                "fold": fold_number,
                "r2": r2,
                "mae": mae,
                "spearman": spearman,
            })

comparison_cv_results = pd.DataFrame(
    comparison_rows
)

print("\nFinished 128D versus 130D comparison.")

display(comparison_cv_results.head())
```

```python
comparison_summary = (
    comparison_cv_results
    .groupby(
        [
            "feature_set",
            "minimum_profiles",
            "n_compounds",
        ],
        as_index=False,
    )
    .agg(
        mean_r2=("r2", "mean"),
        std_r2=("r2", "std"),
        mean_spearman=("spearman", "mean"),
        std_spearman=("spearman", "std"),
        mean_mae=("mae", "mean"),
        std_mae=("mae", "std"),
    )
    .sort_values(
        [
            "minimum_profiles",
            "feature_set",
        ]
    )
    .reset_index(drop=True)
)

display(comparison_summary)
```

```python
comparison_wide = comparison_summary.pivot(
    index=[
        "minimum_profiles",
        "n_compounds",
    ],
    columns="feature_set",
    values=[
        "mean_r2",
        "mean_spearman",
        "mean_mae",
    ],
)

comparison_wide.columns = [
    "_".join(col)
    for col in comparison_wide.columns
]

comparison_wide = (
    comparison_wide
    .reset_index()
)

comparison_wide["delta_r2_130_minus_128"] = (
    comparison_wide[
        "mean_r2_130D_embedding_plus_step1_QC"
    ]
    - comparison_wide[
        "mean_r2_128D_embedding_only"
    ]
)

comparison_wide["delta_spearman_130_minus_128"] = (
    comparison_wide[
        "mean_spearman_130D_embedding_plus_step1_QC"
    ]
    - comparison_wide[
        "mean_spearman_128D_embedding_only"
    ]
)

display(comparison_wide)
```

```python
from sklearn.ensemble import ExtraTreesRegressor
from sklearn.metrics import mean_absolute_error, r2_score
from sklearn.model_selection import RepeatedKFold
from scipy.stats import spearmanr

RANDOM_STATE = 42

# Use the cleaner representation
X_final = X_all

profile_counts = (
    compound_table["n_profiles"]
    .to_numpy()
)

# Only high-confidence compounds in the test sets
high_confidence_indices = np.where(
    profile_counts >= 100
)[0]

print(
    "High-confidence compounds available for testing:",
    len(high_confidence_indices)
)

training_thresholds = [5, 10, 20, 50, 75, 100]

cv = RepeatedKFold(
    n_splits=5,
    n_repeats=5,
    random_state=RANDOM_STATE,
)

fixed_test_rows = []

for fold_number, (high_train_local, high_test_local) in enumerate(
    cv.split(high_confidence_indices),
    start=1,
):
    # Row positions in compound_table/X_final
    test_indices = high_confidence_indices[high_test_local]

    # Eval on this same test set
    X_test = X_final[test_indices]
    y_test = y_all[test_indices]

    for threshold in training_thresholds:
        eligible_indices = np.where(
            profile_counts >= threshold
        )[0]

        # Remove the fixed high-confidence test compounds from training
        train_indices = np.setdiff1d(
            eligible_indices,
            test_indices,
            assume_unique=False,
        )

        model = ExtraTreesRegressor(
            n_estimators=400,
            max_depth=None,
            min_samples_leaf=3,
            random_state=RANDOM_STATE + fold_number,
            n_jobs=-1,
        )

        model.fit(
            X_final[train_indices],
            y_all[train_indices],
        )

        predictions = model.predict(X_test)

        r2 = r2_score(
            y_test,
            predictions,
        )

        mae = mean_absolute_error(
            y_test,
            predictions,
        )

        spearman = spearmanr(
            y_test,
            predictions,
        ).correlation

        if np.isnan(spearman):
            spearman = 0.0

        fixed_test_rows.append({
            "fold": fold_number,
            "training_minimum_profiles": threshold,
            "n_training_compounds": len(train_indices),
            "n_test_compounds": len(test_indices),
            "r2": r2,
            "spearman": spearman,
            "mae": mae,
        })

fixed_high_conf_test_results = pd.DataFrame(
    fixed_test_rows
)

print("Finished fixed-test-set experiment.")

display(fixed_high_conf_test_results.head())
```

```python
fixed_high_conf_summary = (
    fixed_high_conf_test_results
    .groupby(
        [
            "training_minimum_profiles",
            "n_training_compounds",
        ],
        as_index=False,
    )
    .agg(
        mean_r2=("r2", "mean"),
        std_r2=("r2", "std"),
        mean_spearman=("spearman", "mean"),
        std_spearman=("spearman", "std"),
        mean_mae=("mae", "mean"),
        std_mae=("mae", "std"),
    )
    .sort_values("training_minimum_profiles")
    .reset_index(drop=True)
)

display(fixed_high_conf_summary)
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.linear_model import Ridge
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import PolynomialFeatures, StandardScaler

profile_count_feature = np.log1p(
    compound_table["n_profiles"].to_numpy(dtype=np.float32)
).reshape(-1, 1)

observed_predictability = compound_table[
    "source_normalized_inverted_score"
].to_numpy(dtype=np.float32)

print("Profile-count feature shape:", profile_count_feature.shape)
print("Predictability target shape:", observed_predictability.shape)

print(
    "Spearman correlation between profile count and predictability:",
    compound_table["n_profiles"].corr(
        compound_table["source_normalized_inverted_score"],
        method="spearman",
    ),
)
```

```python
import numpy as np
import pandas as pd

from sklearn.ensemble import ExtraTreesRegressor
from sklearn.linear_model import Ridge
from sklearn.metrics import (
    mean_absolute_error,
    r2_score,
)
from sklearn.model_selection import RepeatedKFold
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import (
    PolynomialFeatures,
    StandardScaler,
)
from scipy.stats import spearmanr

RANDOM_STATE = 42

X_structure = X_all

profile_counts = (
    compound_table["n_profiles"]
    .to_numpy(dtype=np.float32)
)

raw_scores = (
    compound_table["source_normalized_inverted_score"]
    .to_numpy(dtype=np.float32)
)

log_profile_counts = np.log1p(
    profile_counts
).reshape(-1, 1)

training_thresholds = [
    5,
    10,
    20,
    50,
    75,
    100,
]

# The test sets always contain only high-confidence compounds
high_confidence_indices = np.where(
    profile_counts >= 100
)[0]

print(
    "High-confidence compounds available for testing:",
    len(high_confidence_indices),
)

cv = RepeatedKFold(
    n_splits=5,
    n_repeats=5,
    random_state=RANDOM_STATE,
)

adjusted_result_rows = []

for fold_number, (
    high_train_local,
    high_test_local,
) in enumerate(
    cv.split(high_confidence_indices),
    start=1,
):
    test_indices = high_confidence_indices[
        high_test_local
    ]

    for threshold in training_thresholds:
        eligible_indices = np.where(
            profile_counts >= threshold
        )[0]

        # Prevent held-out compounds from entering training
        train_indices = np.setdiff1d(
            eligible_indices,
            test_indices,
            assume_unique=False,
        )

        # Fit expected predictability from profile count
        # using training compounds only
        adjustment_model = make_pipeline(
            PolynomialFeatures(
                degree=2,
                include_bias=False,
            ),
            StandardScaler(),
            Ridge(alpha=1.0),
        )

        adjustment_model.fit(
            log_profile_counts[train_indices],
            raw_scores[train_indices],
        )

        expected_train_scores = adjustment_model.predict(
            log_profile_counts[train_indices]
        )

        expected_test_scores = adjustment_model.predict(
            log_profile_counts[test_indices]
        )

        adjusted_train_scores = (
            raw_scores[train_indices]
            - expected_train_scores
        )

        adjusted_test_scores = (
            raw_scores[test_indices]
            - expected_test_scores
        )

        # Train the structure-only Step 4 model
        model = ExtraTreesRegressor(
            n_estimators=400,
            max_depth=None,
            min_samples_leaf=3,
            random_state=RANDOM_STATE + fold_number,
            n_jobs=-1,
        )

        model.fit(
            X_structure[train_indices],
            adjusted_train_scores,
        )

        predictions = model.predict(
            X_structure[test_indices]
        )

        r2 = r2_score(
            adjusted_test_scores,
            predictions,
        )

        mae = mean_absolute_error(
            adjusted_test_scores,
            predictions,
        )

        spearman = spearmanr(
            adjusted_test_scores,
            predictions,
        ).correlation

        if np.isnan(spearman):
            spearman = 0.0

        remaining_profile_correlation = spearmanr(
            profile_counts[train_indices],
            adjusted_train_scores,
        ).correlation

        adjusted_result_rows.append({
            "fold": fold_number,
            "training_minimum_profiles": threshold,
            "n_training_compounds": len(train_indices),
            "n_test_compounds": len(test_indices),
            "r2": r2,
            "spearman": spearman,
            "mae": mae,
            "remaining_train_profile_correlation":
                remaining_profile_correlation,
        })

adjusted_fixed_test_results = pd.DataFrame(
    adjusted_result_rows
)

print("Finished profile-count-adjusted experiment.")

display(adjusted_fixed_test_results.head())
```

```python
# Applicability-domain analysis
# Step 2: recreate the exact held-out split and calculate
# Morgan/Tanimoto domain-support metrics

from scipy.stats import spearmanr



# Load the complete n_profiles >= 50 compound table


FINAL_TRAINING_TABLE_PATH = (
    STEP3_DIR
    / "step4_final_extra_trees_n50_training_compounds.csv"
)

assert FINAL_TRAINING_TABLE_PATH.exists(), (
    f"Missing final training table: {FINAL_TRAINING_TABLE_PATH}"
)

all_n50_compounds = pd.read_csv(
    FINAL_TRAINING_TABLE_PATH,
    low_memory=False,
)

print("All eligible compounds:", len(all_n50_compounds))
print("Columns:", all_n50_compounds.columns.tolist())

assert "compound_embedding_row_index" in all_n50_compounds.columns



# Recover the exact 382-training / 96-test split


all_indices = (
    all_n50_compounds["compound_embedding_row_index"]
    .astype(int)
    .to_numpy()
)

test_indices = (
    holdout_predictions["compound_embedding_row_index"]
    .astype(int)
    .to_numpy()
)

test_index_set = set(test_indices.tolist())

train_indices = np.array(
    [
        index
        for index in all_indices
        if index not in test_index_set
    ],
    dtype=int,
)

print("\nRecovered split")
print("Training compounds:", len(train_indices))
print("Held-out compounds:", len(test_indices))
print(
    "Overlap:",
    len(set(train_indices).intersection(test_index_set)),
)

assert len(all_indices) == 478
assert len(train_indices) == 382
assert len(test_indices) == 96
assert len(set(train_indices).intersection(test_index_set)) == 0

assert train_indices.max() < morgan_packed.shape[0]
assert test_indices.max() < morgan_packed.shape[0]



# Select and unpack only the required Morgan fingerprints


train_packed = np.asarray(
    morgan_packed[train_indices]
)

test_packed = np.asarray(
    morgan_packed[test_indices]
)

train_bits = np.unpackbits(
    train_packed,
    axis=1,
)

test_bits = np.unpackbits(
    test_packed,
    axis=1,
)

print("\nFingerprint matrices")
print("Training fingerprints:", train_bits.shape)
print("Test fingerprints:", test_bits.shape)

assert train_bits.shape == (382, 2048)
assert test_bits.shape == (96, 2048)



# Calculate all test-to-training Tanimoto similarities


train_bits_int = train_bits.astype(
    np.int32
)

test_bits_int = test_bits.astype(
    np.int32
)

intersections = (
    test_bits_int
    @ train_bits_int.T
)

test_bit_counts = (
    test_bits_int.sum(axis=1)
    [:, None]
)

train_bit_counts = (
    train_bits_int.sum(axis=1)
    [None, :]
)

unions = (
    test_bit_counts
    + train_bit_counts
    - intersections
)

tanimoto_matrix = np.divide(
    intersections,
    unions,
    out=np.zeros_like(
        intersections,
        dtype=float,
    ),
    where=unions != 0,
)

print("\nTanimoto matrix:", tanimoto_matrix.shape)
print(
    "Similarity range:",
    float(tanimoto_matrix.min()),
    "to",
    float(tanimoto_matrix.max()),
)



# Metric 1: nearest training-compound similarity


nearest_train_position = (
    tanimoto_matrix.argmax(axis=1)
)

nearest_tanimoto = (
    tanimoto_matrix.max(axis=1)
)



# Metric 2: mean similarity to five nearest training compounds


top5_tanimoto = np.partition(
    tanimoto_matrix,
    kth=-5,
    axis=1,
)[:, -5:]

mean_top5_tanimoto = (
    top5_tanimoto.mean(axis=1)
)



# Metric 3: Sum of Distance-Weighted Contributions (SDC)
#
# Tanimoto distance = 1 - similarity
#
# Published weighting:
# exp[-3 * distance / (1 - distance)]

sdc_weights = np.zeros_like(
    tanimoto_matrix,
    dtype=float,
)

positive_similarity = (
    tanimoto_matrix > 0
)

sdc_weights[positive_similarity] = np.exp(
    -3.0
    * (
        1.0
        - tanimoto_matrix[positive_similarity]
    )
    / tanimoto_matrix[positive_similarity]
)

sdc = sdc_weights.sum(
    axis=1
)


# Attach the closest training-compound identity


train_name_lookup = (
    all_n50_compounds
    .set_index("compound_embedding_row_index")
)

nearest_training_indices = (
    train_indices[nearest_train_position]
)

nearest_training_names = [
    (
        train_name_lookup.loc[
            index,
            "compound_name",
        ]
        if (
            "compound_name"
            in train_name_lookup.columns
        )
        else ""
    )
    for index in nearest_training_indices
]



# Add the applicability metrics to the held-out predictions


applicability_results = (
    holdout_predictions
    .copy()
)

applicability_results["absolute_error"] = np.abs(
    applicability_results["actual"]
    - applicability_results["predicted"]
)

applicability_results[
    "nearest_training_row_index"
] = nearest_training_indices

applicability_results[
    "nearest_training_compound"
] = nearest_training_names

applicability_results[
    "nearest_tanimoto"
] = nearest_tanimoto

applicability_results[
    "mean_top5_tanimoto"
] = mean_top5_tanimoto

applicability_results[
    "sdc"
] = sdc



# Check whether domain support relates to prediction error


domain_metrics = [
    "nearest_tanimoto",
    "mean_top5_tanimoto",
    "sdc",
]

print("\nAssociation with absolute prediction error")

for metric in domain_metrics:

    correlation = spearmanr(
        applicability_results[metric],
        applicability_results["absolute_error"],
    ).correlation

    print(
        f"{metric:25s}: "
        f"Spearman = {correlation:.3f}"
    )



# Save the results

APPLICABILITY_RESULTS_PATH = (
    STEP3_DIR
    / "step4_extra_trees_n50_holdout_applicability_metrics.csv"
)

applicability_results.to_csv(
    APPLICABILITY_RESULTS_PATH,
    index=False,
)

print("\nSaved applicability results:")
print(APPLICABILITY_RESULTS_PATH)

display(
    applicability_results[
        [
            "compound_name",
            "actual",
            "predicted",
            "absolute_error",
            "nearest_training_compound",
            "nearest_tanimoto",
            "mean_top5_tanimoto",
            "sdc",
        ]
    ]
    .sort_values(
        "nearest_tanimoto"
    )
    .head(10)
)
```

```python
# Applicability-domain analysis
# Step 3: inspect similarity distribution 

import numpy as np
import pandas as pd
from scipy.stats import spearmanr



# Summarize the nearest-neighbor similarity distribution


print("Nearest-training-compound Tanimoto summary")
display(
    applicability_results["nearest_tanimoto"]
    .describe(
        percentiles=[
            0.10,
            0.25,
            0.50,
            0.75,
            0.90,
            0.95,
        ]
    )
    .to_frame("nearest_tanimoto")
)



# Count compounds above common similarity levels


similarity_thresholds = [
    1.00,
    0.90,
    0.80,
    0.70,
    0.60,
    0.50,
    0.40,
    0.30,
    0.20,
]

threshold_summary = pd.DataFrame(
    [
        {
            "minimum_nearest_tanimoto": threshold,
            "n_test_compounds": int(
                (
                    applicability_results[
                        "nearest_tanimoto"
                    ]
                    >= threshold
                ).sum()
            ),
        }
        for threshold in similarity_thresholds
    ]
)

print("\nNumber of test compounds above each similarity level")
display(threshold_summary)



# Identify exact Morgan-fingerprint matches

applicability_results[
    "exact_morgan_fingerprint_match"
] = np.isclose(
    applicability_results["nearest_tanimoto"],
    1.0,
    atol=1e-12,
)

exact_matches = applicability_results.loc[
    applicability_results[
        "exact_morgan_fingerprint_match"
    ]
].copy()

print(
    "\nHeld-out compounds with an exact "
    "Morgan-fingerprint match in training:",
    len(exact_matches),
)

if len(exact_matches) > 0:

    display(
        exact_matches[
            [
                "compound_name",
                "nearest_training_compound",
                "nearest_tanimoto",
                "actual",
                "predicted",
                "absolute_error",
            ]
        ]
        .sort_values(
            "absolute_error",
            ascending=False,
        )
    )



# Recalculate correlations after excluding exact matches


nonduplicate_results = applicability_results.loc[
    ~applicability_results[
        "exact_morgan_fingerprint_match"
    ]
].copy()

print(
    "\nCompounds remaining after excluding exact matches:",
    len(nonduplicate_results),
)

print("\nCorrelation with error after excluding exact matches")

nonduplicate_correlations = []

for metric in domain_metrics:

    correlation = spearmanr(
        nonduplicate_results[metric],
        nonduplicate_results["absolute_error"],
    ).correlation

    nonduplicate_correlations.append(
        {
            "domain_metric": metric,
            "spearman_with_absolute_error": correlation,
            "n_compounds": len(nonduplicate_results),
        }
    )

    print(
        f"{metric:25s}: "
        f"Spearman = {correlation:.3f}"
    )

nonduplicate_correlation_table = pd.DataFrame(
    nonduplicate_correlations
)



# Divide compounds into nearest-similarity quartiles

applicability_results[
    "nearest_similarity_quartile"
] = pd.qcut(
    applicability_results["nearest_tanimoto"],
    q=4,
    duplicates="drop",
)

similarity_quartile_summary = (
    applicability_results
    .groupby(
        "nearest_similarity_quartile",
        observed=True,
    )
    .agg(
        n_compounds=("absolute_error", "size"),
        minimum_similarity=(
            "nearest_tanimoto",
            "min",
        ),
        maximum_similarity=(
            "nearest_tanimoto",
            "max",
        ),
        mean_absolute_error=(
            "absolute_error",
            "mean",
        ),
        median_absolute_error=(
            "absolute_error",
            "median",
        ),
    )
    .reset_index()
)

print("\nPrediction error by similarity quartile")
display(similarity_quartile_summary)



# Save the diagnostic summaries

THRESHOLD_SUMMARY_PATH = (
    STEP3_DIR
    / "step4_n50_applicability_similarity_threshold_counts.csv"
)

QUARTILE_SUMMARY_PATH = (
    STEP3_DIR
    / "step4_n50_applicability_similarity_quartile_errors.csv"
)

NONDUPLICATE_CORRELATION_PATH = (
    STEP3_DIR
    / "step4_n50_applicability_nonduplicate_correlations.csv"
)

threshold_summary.to_csv(
    THRESHOLD_SUMMARY_PATH,
    index=False,
)

similarity_quartile_summary.to_csv(
    QUARTILE_SUMMARY_PATH,
    index=False,
)

nonduplicate_correlation_table.to_csv(
    NONDUPLICATE_CORRELATION_PATH,
    index=False,
)

print("\nSaved diagnostic summaries:")
print(THRESHOLD_SUMMARY_PATH)
print(QUARTILE_SUMMARY_PATH)
print(NONDUPLICATE_CORRELATION_PATH)
```

```python
# Applicability-domain analysis
# Step 4: visualize Morgan similarity versus prediction error

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import spearmanr


# Figure path


FIG_DIR = (
    PROJECT_ROOT
    / "figures"
    / "step4_predictability_regression"
)

FIG_DIR.mkdir(
    parents=True,
    exist_ok=True,
)

MORGAN_DOMAIN_FIGURE_PATH = (
    FIG_DIR
    / "step4_n50_morgan_similarity_vs_absolute_error.png"
)



# Use compounds without exact fingerprint matches


plot_data = nonduplicate_results.copy()

plot_spearman = spearmanr(
    plot_data["nearest_tanimoto"],
    plot_data["absolute_error"],
).correlation


# Assign similarity quartiles


plot_data["similarity_quartile"] = pd.qcut(
    plot_data["nearest_tanimoto"],
    q=4,
    labels=[
        "Lowest similarity",
        "Low–medium",
        "Medium–high",
        "Highest similarity",
    ],
)

quartile_plot_summary = (
    plot_data
    .groupby(
        "similarity_quartile",
        observed=True,
    )
    .agg(
        mean_similarity=(
            "nearest_tanimoto",
            "mean",
        ),
        mean_absolute_error=(
            "absolute_error",
            "mean",
        ),
        standard_error=(
            "absolute_error",
            lambda values: (
                values.std(ddof=1)
                / np.sqrt(len(values))
            ),
        ),
        n_compounds=(
            "absolute_error",
            "size",
        ),
    )
    .reset_index()
)



# Create figure


fig, ax = plt.subplots(
    figsize=(11, 8)
)

ax.scatter(
    plot_data["nearest_tanimoto"],
    plot_data["absolute_error"],
    alpha=0.55,
    s=65,
    label="Held-out compounds",
)

ax.errorbar(
    quartile_plot_summary["mean_similarity"],
    quartile_plot_summary["mean_absolute_error"],
    yerr=quartile_plot_summary["standard_error"],
    marker="o",
    linestyle="-",
    linewidth=2.5,
    markersize=10,
    capsize=6,
    label="Quartile mean ± standard error",
)

ax.set_title(
    "Morgan Similarity Does Not Strongly Explain Prediction Error",
    fontsize=23,
    pad=18,
)

ax.set_xlabel(
    "Tanimoto similarity to nearest training compound",
    fontsize=18,
    labelpad=10,
)

ax.set_ylabel(
    "Absolute predictability-score error",
    fontsize=18,
    labelpad=10,
)

ax.tick_params(
    axis="both",
    labelsize=15,
)

ax.legend(
    fontsize=14,
)

fig.text(
    0.5,
    0.02,
    (
        f"Nonduplicate held-out compounds: {len(plot_data)} | "
        f"Spearman = {plot_spearman:.3f}"
    ),
    ha="center",
    fontsize=13,
)

fig.subplots_adjust(
    left=0.13,
    right=0.97,
    top=0.88,
    bottom=0.14,
)

plt.savefig(
    MORGAN_DOMAIN_FIGURE_PATH,
    dpi=300,
    bbox_inches="tight",
)

plt.show()



# Print the difference between the extreme quartiles

lowest_quartile_error = (
    quartile_plot_summary
    .iloc[0]["mean_absolute_error"]
)

highest_quartile_error = (
    quartile_plot_summary
    .iloc[-1]["mean_absolute_error"]
)

print(
    "Lowest-similarity quartile mean error:",
    round(lowest_quartile_error, 4),
)

print(
    "Highest-similarity quartile mean error:",
    round(highest_quartile_error, 4),
)

print(
    "Difference:",
    round(
        lowest_quartile_error
        - highest_quartile_error,
        4,
    ),
)

print("\nSaved figure:")
print(MORGAN_DOMAIN_FIGURE_PATH)

display(quartile_plot_summary)
```
